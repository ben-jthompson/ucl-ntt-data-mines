```python
import geopandas as gpd
import pandas as pd
```

```python
england = gpd.read_file('../../data/geojson/local-authority-district.geojson')
wales = gpd.read_file('../../data/geojson/local-authority-district-wales.geojson')
scotland = gpd.read_file('../../data/geojson/local-authority-district-scotland.geojson')
```

```python
scotland = scotland.rename(columns={'local_authority': 'name'})[['name', 'geometry']]
wales = wales.rename(columns={'name_en': 'name'})[['name', 'geometry']]
england = england[['name', 'geometry']].to_crs('EPSG:27700')
```

```python
uk_authorities = gpd.GeoDataFrame(
    pd.concat([england, wales, scotland], ignore_index=True),
    crs=england.crs
)
```

```python
uk_authorities.to_file("../../data/geojson/uk-local-authorities.geojson", driver="GeoJSON")
```
